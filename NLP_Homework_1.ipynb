## 1. Tokenize a Paragraph

హైదరాబాద్‌లో నిన్న భారీ వర్షం పడింది.  
ప్రజలు ట్రాఫిక్‌కు ఇబ్బంది పడ్డారు, కానీ మెట్రోతో త్వరగా ఇంటికి చేరుకున్నారు.  
ఉదయం నుంచే వాతావరణం చల్లగా ఉంది.  
ఇలాంటి రోజుల్లో బయటికి వెళ్లే ముందు జాగ్రత్తలు తీసుకోవాలి.



```python
text = """హైదరాబాద్‌లో నిన్న భారీ వర్షం పడింది.
ప్రజలు ట్రాఫిక్‌కు ఇబ్బంది పడ్డారు, కానీ మెట్రోతో త్వరగా ఇంటికి చేరుకున్నారు.
ఉదయం నుంచే వాతావరణం చల్లగా ఉంది.
ఇలాంటి రోజుల్లో బయటికి వెళ్లే ముందు జాగ్రత్తలు తీసుకోవాలి."""

naive_tokens = text.split()
print("Naive Tokens:")
print(naive_tokens)
```

```python
manual_tokens = [
    'నిన్న', 'హైదరాబాద్', 'లో', 'భారీ', 'వర్షం',
    'పడింది', '.',
    'ప్రజలు', 'ట్రాఫిక్', 'కు', 'ఇబ్బంది',
    'పడ్డారు', '.',
    'మెట్రో', 'రైలు', 'తో', 'చాలామంది',
    'త్వరగా', 'ఇంటికి', 'చేరుకున్నారు', '.',
    'వాతావరణం', 'చల్లగా', 'ఉండటంతో',
    'విద్యార్థులు', 'సంతోషించారు', '.',
    'అయితే', 'రోడ్ల', 'పై',
    'నీరు', 'నిలిచిందని', 'వాహనదారులు',
    'చెప్పారు', '.'
]

print(manual_tokens)
```

2. Comparison of Tokenization Methods    

Let's compare the `naive_tokens` (simple split), `manual_tokens`, and `indic_tokens`.

```python
from indicnlp.tokenize import indic_tokenize

indic_tokens = indic_tokenize.trivial_tokenize(text, lang="te")
print("Naive :", naive_tokens)
print("Manual:", manual_tokens)
print("Indic :", indic_tokens)

# Compare Naive vs Indic
naive_only = sorted(set(naive_tokens) - set(indic_tokens))
indic_only_from_naive = sorted(set(indic_tokens) - set(naive_tokens))

print("\n=== Naive vs Indic Comparison ===")
print("Tokens only in Naive:", naive_only)
print("Tokens only in Indic:", indic_only_from_naive)

# Compare Manual vs Indic
manual_only = sorted(set(manual_tokens) - set(indic_tokens))
indic_only_from_manual = sorted(set(indic_tokens) - set(manual_tokens))

print("\n=== Manual vs Indic Comparison ===")
print("Tokens only in Manual:", manual_only)
print("Tokens only in Indic:", indic_only_from_manual)

sym_naive_indic = set(naive_tokens) ^ set(indic_tokens)
sym_manual_indic = set(manual_tokens) ^ set(indic_tokens)

print("\n=== Symmetric Differences (summary) ===")
print("Naive ↔ Indic :", sym_naive_indic)
print("Manual ↔ Indic:", sym_manual_indic)
```

Analysis of Differences

The indic_nlp_library tokenizer generates more linguistically consistent tokens than simple space-based splitting because it follows language-specific rules. The main differences are:

Punctuation handling: In naïve tokenization using split(), punctuation tends to remain attached to words (for example, forms like పడింది., పడ్డారు,, or చెప్పారు.). The Indic NLP tokenizer, however, generally separates punctuation marks such as . and , into independent tokens, making the output cleaner and easier to process for NLP tasks.

Handling of suffixes and word forms: Telugu frequently attaches case markers or postpositions to words (e.g., లో, కు, తో, పై). In my manual tokenization, I explicitly separated these endings from the root words (e.g., హైదరాబాద్ + లో, ట్రాఫిక్ + కు, మెట్రో + తో). The Indic tokenizer may either keep such forms attached or segment them differently based on its internal rules, which leads to visible differences between manual_tokens and indic_tokens.

### 3. Multiword Expressions (MWEs)

Multiword Expressions (MWEs) are groups of words that together express a single meaning that cannot be fully understood by interpreting each word separately. In NLP, treating them as single tokens is important because splitting them can distort meaning and reduce model accuracy.

Below are three MWEs from your Hyderabad rain paragraph:

1) హైదరాబాద్ నగరం (Hyderabad City)

Although హైదరాబాద్ means the place name and నగరం means “city,” together they refer to a specific urban location.

If treated as separate tokens, an NLP system may lose the named-entity meaning of this specific city.

2) ట్రాఫిక్ జామ్ (Traffic Jam)

Individually, ట్రాఫిక్ means “traffic” and జామ్ means “blockage,” but together they form a fixed phrase meaning road congestion.

Splitting them would weaken the intended meaning of a real-world situation that commonly occurs during heavy rain.

3) మెట్రో రైలు (Metro Train)

మెట్రో refers to the rapid transit system and రైలు means “train,” but together they denote a specific transportation system in Hyderabad.

Treating them separately could confuse models that analyze transportation or urban mobility.

### 4. Reflection

Tokenization in Telugu proved to be particularly challenging because of its highly agglutinative structure and rich morphology. Unlike English, where words are mostly separated by spaces and contain fewer inflections, Telugu words often carry multiple suffixes attached to a single root, making it difficult to determine clear token boundaries. This makes Telugu tokenization considerably more complex than English, where simple whitespace and punctuation rules often suffice.

Punctuation, morphology, and Multiword Expressions (MWEs) together make tokenization more demanding. Telugu punctuation is frequently attached to words, requiring careful separation so that meaning is not lost. Additionally, the language’s complex word formation means that improper splitting can distort the intended sense of a sentence. Finally, MWEs function as single semantic units, so treating them as separate tokens can break their meaning, requiring more thoughtful and language-aware tokenization strategies for effective NLP processing.
